```python
# ============================================================
# EXPERIMENT 6: ROLE-BASED AND NEGATIVE PROMPTING
# Groq API + Google Colab
# ============================================================

!pip install -q groq

import os
from getpass import getpass
from groq import Groq

# ------------------------------------------------------------
# STEP 1: Load Groq API Key
# ------------------------------------------------------------

groq_api_key = getpass("Enter your Groq API key: ")

if not groq_api_key:
    raise ValueError("GROQ_API_KEY not provided.")

os.environ["GROQ_API_KEY"] = groq_api_key

print("✅ Groq API key loaded successfully.")


# ------------------------------------------------------------
# STEP 2: Create Groq Client
# ------------------------------------------------------------

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

print("✅ Groq client initialized.")


# ------------------------------------------------------------
# STEP 3: Define Function
# ------------------------------------------------------------

def ask_llm(prompt):

    try:
        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            temperature=0.5,
            max_tokens=200,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        return response.choices[0].message.content.strip()

    except Exception as e:
        return f"❌ Error: {e}"


# ------------------------------------------------------------
# STEP 4: Define Prompts
# ------------------------------------------------------------

baseline_prompt = (
    "Explain the importance of exercise."
)

role_prompt = (
    "You are a motivational fitness coach. "
    "Explain the importance of exercise."
)

role_negative_prompt = (
    "You are a motivational fitness coach. "
    "Do not use any medical jargon. "
    "Do not mention weight loss. "
    "Explain the importance of exercise."
)


# ------------------------------------------------------------
# STEP 5: Generate Responses
# ------------------------------------------------------------

baseline_response = ask_llm(baseline_prompt)

role_response = ask_llm(role_prompt)

role_negative_response = ask_llm(
    role_negative_prompt
)


# ============================================================
# STEP 6: Display BASELINE
# ============================================================

print("\n" + "=" * 70)
print("BASELINE")
print("=" * 70)

print("\nPrompt:")
print(baseline_prompt)

print("\nResponse:")
print(baseline_response)


# ============================================================
# STEP 7: Display ROLE-BASED
# ============================================================

print("\n" + "=" * 70)
print("ROLE-BASED ONLY")
print("=" * 70)

print("\nPrompt:")
print(role_prompt)

print("\nResponse:")
print(role_response)


# ============================================================
# STEP 8: Display ROLE + NEGATIVE
# ============================================================

print("\n" + "=" * 70)
print("ROLE + NEGATIVE PROMPTING")
print("=" * 70)

print("\nPrompt:")
print(role_negative_prompt)

print("\nResponse:")
print(role_negative_response)


# ============================================================
# STEP 9: Check Negative Constraints
# ============================================================

response_lower = role_negative_response.lower()

weight_loss_words = [
    "weight loss",
    "lose weight",
    "losing weight",
    "weight reduction"
]

medical_jargon_words = [
    "cardiovascular",
    "metabolism",
    "endocrine",
    "hypertension",
    "aerobic capacity",
    "musculoskeletal"
]


weight_loss_found = any(
    word in response_lower
    for word in weight_loss_words
)

medical_jargon_found = any(
    word in response_lower
    for word in medical_jargon_words
)


# ============================================================
# STEP 10: Analysis
# ============================================================

print("\n" + "=" * 70)
print("ANALYSIS")
print("=" * 70)

print("\n1. Tone:")
print(
    "The baseline response is generally neutral and informative."
)

print(
    "The role-based response should sound more motivational "
    "and encouraging because the model is instructed to act "
    "as a fitness coach."
)

print(
    "The role + negative response should maintain the "
    "motivational tone while avoiding the specified topics."
)


print("\n2. Vocabulary:")

print(
    "The baseline may use general explanatory vocabulary."
)

print(
    "The fitness-coach role may produce more energetic, "
    "encouraging, and action-oriented vocabulary."
)

print(
    "The negative prompt should reduce the use of medical "
    "terminology and avoid weight-loss-related language."
)


print("\n3. Constraint Adherence:")

if weight_loss_found:
    print(
        "❌ Weight-loss constraint violated: "
        "weight-loss-related wording was detected."
    )
else:
    print(
        "✅ Weight-loss constraint respected."
    )


if medical_jargon_found:
    print(
        "⚠️ Possible medical jargon detected."
    )
else:
    print(
        "✅ No predefined medical jargon detected."
    )


# ============================================================
# STEP 11: Final Summary
# ============================================================

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print("""
Baseline:
Provides a general explanation without a specified persona.

Role-Based:
The fitness-coach role encourages a more motivational,
friendly, and action-oriented response.

Role + Negative:
The response should retain the motivational fitness-coach
style while avoiding medical jargon and weight-loss discussion.

Conclusion:
Role prompting can influence the tone and perspective of an
LLM response, while negative prompting can restrict unwanted
content when the constraints are clear and specific.
""")

print("✅ Experiment 6 completed successfully.")
```
