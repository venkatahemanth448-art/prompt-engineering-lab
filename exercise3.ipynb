```python
# ============================================================
# EXPERIMENT 3: ITERATIVE PROMPT REFINEMENT
# Topic: Explain Photosynthesis to a 10-Year-Old
# Groq API + Google Colab
# ============================================================

import os
from getpass import getpass
from groq import Groq

# ------------------------------------------------------------
# STEP 1: Load API Key
# ------------------------------------------------------------

groq_api_key = getpass("Enter your Groq API key: ")

if not groq_api_key:
    raise ValueError("GROQ_API_KEY not provided.")

os.environ["GROQ_API_KEY"] = groq_api_key

print("✅ Groq API key loaded successfully.")


# ------------------------------------------------------------
# STEP 2: Create Groq Client
# ------------------------------------------------------------

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

print("✅ Groq client initialized.")


# ------------------------------------------------------------
# STEP 3: Define LLM Function
# ------------------------------------------------------------

def ask_llm(prompt):
    try:

        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            temperature=0.5,
            max_tokens=250,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"❌ Error: {e}"


# ------------------------------------------------------------
# STEP 4: Iteration 0 - Baseline
# ------------------------------------------------------------

prompt0 = (
    "Explain photosynthesis to a 10-year-old."
)


# ------------------------------------------------------------
# STEP 5: Iteration 1 - First Refinement
# ------------------------------------------------------------

prompt1 = (
    "You are a friendly science teacher. "
    "Explain photosynthesis to a 10-year-old using a simple analogy. "
    "Keep the explanation to 3 short sentences and avoid technical jargon."
)


# ------------------------------------------------------------
# STEP 6: Iteration 2 - Second Refinement
# ------------------------------------------------------------

prompt2 = (
    "You are a friendly science teacher. "
    "Explain photosynthesis to a 10-year-old using a simple analogy. "
    "Keep it to 3 short sentences, avoid technical jargon, "
    "include a simple memory trick and describe a simple diagram."
)


# ------------------------------------------------------------
# STEP 7: Store Prompts
# ------------------------------------------------------------

prompts = [
    prompt0,
    prompt1,
    prompt2
]

responses = []


# ------------------------------------------------------------
# STEP 8: Execute All Iterations
# ------------------------------------------------------------

for i, prompt in enumerate(prompts):

    print("\n" + "=" * 70)
    print(f"ITERATION {i}")
    print("=" * 70)

    print("\nPrompt:")
    print(prompt)

    response = ask_llm(prompt)

    responses.append(response)

    print("\nResponse:")
    print(response)

    # --------------------------------------------------------
    # Analysis
    # --------------------------------------------------------

    print("\nAnalysis:")

    if i == 0:

        print(
            "The baseline prompt gives a general explanation "
            "without specifying audience, style, length, or format."
        )

    elif i == 1:

        print(
            "The prompt was refined by adding a teacher role, "
            "a simple analogy, a three-sentence limit, and "
            "a restriction against technical jargon."
        )

    else:

        print(
            "The final refinement adds a memory trick and "
            "diagram description to improve memorability "
            "and visualization."
        )


# ------------------------------------------------------------
# STEP 9: Final Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print("\nIteration 0:")
print("Basic explanation with no additional constraints.")

print("\nIteration 1:")
print(
    "Improved clarity using a role, analogy, simple language, "
    "and a length constraint."
)

print("\nIteration 2:")
print(
    "Further improved learning by adding a memory trick "
    "and a simple diagram description."
)

print("\n✅ Iterative prompt refinement experiment completed.")
```
