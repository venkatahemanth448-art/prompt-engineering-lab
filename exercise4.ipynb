```python
# ============================================================
# EXPERIMENT 4: PROMPT FAILURE DIAGNOSIS
# AND EDGE-CASE HANDLING
# Groq API + Google Colab
# ============================================================

import os
from getpass import getpass
from groq import Groq

# ------------------------------------------------------------
# STEP 1: Load API Key
# ------------------------------------------------------------

groq_api_key = getpass("Enter your Groq API key: ")

if not groq_api_key:
    raise ValueError("GROQ_API_KEY not provided.")

os.environ["GROQ_API_KEY"] = groq_api_key

print("✅ Groq API key loaded successfully.")


# ------------------------------------------------------------
# STEP 2: Create Groq Client
# ------------------------------------------------------------

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

print("✅ Groq client initialized.")


# ------------------------------------------------------------
# STEP 3: Define Function
# ------------------------------------------------------------

def run_prompt(prompt):

    try:
        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            temperature=0.3,
            max_tokens=250,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"❌ API Error: {e}"


# ============================================================
# CASE 1: HALLUCINATION / FACTUAL ERROR
# ============================================================

original_prompt_A = "Who won the FIFA World Cup in 1942?"

corrected_prompt_A = """
The FIFA World Cup was not held in 1942 because of World War II.
Explain why there was no tournament in 1942, and identify the
winner of the previous FIFA World Cup held in 1938.
Do not invent a winner for 1942.
"""


print("\n" + "=" * 70)
print("CASE 1: HALLUCINATION / FACTUAL ERROR")
print("=" * 70)

print("\nOriginal Prompt:")
print(original_prompt_A)

response_A = run_prompt(original_prompt_A)

print("\nOriginal Response:")
print(response_A)

print("\nDiagnosis:")
print("Failure Type: Hallucination / factual error")
print("Cause: The prompt assumes that a World Cup was held in 1942.")
print("The model may incorrectly accept the false premise.")

print("\nCorrected Prompt:")
print(corrected_prompt_A)

corrected_response_A = run_prompt(corrected_prompt_A)

print("\nImproved Response:")
print(corrected_response_A)


# ============================================================
# CASE 2: INSTRUCTION IGNORING / FORMAT CONSTRAINT
# ============================================================

original_prompt_B = (
    "List the three largest oceans on Earth. "
    "Respond using exactly three bullet points, "
    "with each bullet point on a new line starting with a dash."
)

corrected_prompt_B = """
List exactly three of Earth's largest oceans.

Formatting requirements:
- Use exactly three lines.
- Each line must start with "-".
- Include only one ocean name on each line.
- Do not add an introduction or explanation.
"""


print("\n" + "=" * 70)
print("CASE 2: INSTRUCTION IGNORING / FORMAT CONSTRAINT")
print("=" * 70)

print("\nOriginal Prompt:")
print(original_prompt_B)

response_B = run_prompt(original_prompt_B)

print("\nOriginal Response:")
print(response_B)

print("\nDiagnosis:")
print("Failure Type: Instruction ignoring / formatting failure")
print("Cause: The model may add explanations or fail to follow")
print("the exact requested output structure.")

print("\nCorrected Prompt:")
print(corrected_prompt_B)

corrected_response_B = run_prompt(corrected_prompt_B)

print("\nImproved Response:")
print(corrected_response_B)


# ============================================================
# CASE 3: OVER-REFUSAL / SAFETY MIS-FIRE
# ============================================================

original_prompt_C = (
    "Tell me a fun fact about the Roman Empire "
    "that is appropriate for school children."
)

corrected_prompt_C = """
Provide exactly one educational and school-appropriate fun fact
about the Roman Empire. Keep it under 40 words. Do not discuss
violence or sensitive topics.
"""


print("\n" + "=" * 70)
print("CASE 3: OVER-REFUSAL / SAFETY MIS-FIRE")
print("=" * 70)

print("\nOriginal Prompt:")
print(original_prompt_C)

response_C = run_prompt(original_prompt_C)

print("\nOriginal Response:")
print(response_C)

print("\nDiagnosis:")
print("Failure Type: Possible over-refusal")
print("Cause: A model may incorrectly interpret a historical topic")
print("as requiring refusal even though the request is educational.")

print("\nCorrected Prompt:")
print(corrected_prompt_C)

corrected_response_C = run_prompt(corrected_prompt_C)

print("\nImproved Response:")
print(corrected_response_C)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print("""
1. Hallucination:
   The corrected prompt explicitly prevents accepting a false
   premise and asks for historically valid information.

2. Instruction Ignoring:
   The corrected prompt clearly separates the task from the
   formatting requirements and specifies the exact structure.

3. Over-Refusal:
   The corrected prompt makes the educational and harmless
   nature of the request explicit and avoids sensitive topics.

Overall Lesson:
Prompt failures can often be reduced by identifying the exact
failure mode and adding targeted instructions rather than simply
making the prompt longer.
""")

print("✅ Prompt failure diagnosis experiment completed.")
```
