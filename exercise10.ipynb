```python
# ============================================================
# EXPERIMENT 10: REASONING PROMPTING
# Direct Prompt vs Step-by-Step Solution Prompt
# Groq API + Google Colab
# ============================================================

!pip install -q groq

import os
import re
from getpass import getpass
from groq import Groq


# ------------------------------------------------------------
# STEP 1: Load Groq API Key
# ------------------------------------------------------------

groq_api_key = getpass("Enter your Groq API key: ")

if not groq_api_key:
    raise ValueError("GROQ_API_KEY not provided.")

os.environ["GROQ_API_KEY"] = groq_api_key

print("✅ Groq API key loaded successfully.")


# ------------------------------------------------------------
# STEP 2: Create Groq Client
# ------------------------------------------------------------

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

print("✅ Groq client initialized.")


# ------------------------------------------------------------
# STEP 3: Function to Call LLM
# ------------------------------------------------------------

def ask_llm(prompt):

    try:
        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            temperature=0,
            max_tokens=300,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        return response.choices[0].message.content.strip()

    except Exception as e:
        return f"ERROR: {e}"


# ============================================================
# STEP 4: Define the Math Problem
# ============================================================

problem = """
A farmer has 15 cows. He buys 7 more cows.
Then he sells half of his cows.
How many cows does he have left?
"""


# ============================================================
# STEP 5: Direct Prompt
# ============================================================

direct_prompt = problem


# ============================================================
# STEP 6: Step-by-Step Prompt
# ============================================================

step_prompt = f"""
Solve the following math problem.

{problem}

Give a brief calculation summary showing the key arithmetic
steps, then clearly state the final answer.
"""


# ============================================================
# STEP 7: Send Direct Prompt
# ============================================================

print("\n" + "=" * 70)
print("DIRECT PROMPT")
print("=" * 70)

print("\nPrompt:")
print(direct_prompt)

direct_response = ask_llm(direct_prompt)

print("\nResponse:")
print(direct_response)


# ============================================================
# STEP 8: Send Step-by-Step Prompt
# ============================================================

print("\n" + "=" * 70)
print("STEP-BY-STEP PROMPT")
print("=" * 70)

print("\nPrompt:")
print(step_prompt)

step_response = ask_llm(step_prompt)

print("\nResponse:")
print(step_response)


# ============================================================
# STEP 9: Expected Answer
# ============================================================

expected_answer = 11


# ============================================================
# STEP 10: Extract Final Number
# ============================================================

def extract_answer(text):

    # Look for common final-answer patterns
    patterns = [
        r"final answer\s*[:\-]?\s*(\d+)",
        r"answer\s*[:\-]?\s*(\d+)",
        r"(\d+)\s*cows"
    ]

    for pattern in patterns:

        matches = re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        if matches:
            return int(matches[-1])

    # Fallback: take the last number
    numbers = re.findall(r"\b\d+\b", text)

    if numbers:
        return int(numbers[-1])

    return None


direct_answer = extract_answer(direct_response)
step_answer = extract_answer(step_response)


# ============================================================
# STEP 11: Compare Answers
# ============================================================

print("\n" + "=" * 70)
print("COMPARISON")
print("=" * 70)

print(f"\nExpected answer: {expected_answer}")

print(f"Direct prompt answer: {direct_answer}")

if direct_answer == expected_answer:
    print("✅ Direct prompt is correct.")
else:
    print("❌ Direct prompt is incorrect.")


print(f"\nStep-by-step answer: {step_answer}")

if step_answer == expected_answer:
    print("✅ Step-by-step prompt is correct.")
else:
    print("❌ Step-by-step prompt is incorrect.")


# ============================================================
# STEP 12: Optional Decomposition
# ============================================================

print("\n" + "=" * 70)
print("OPTIONAL TASK DECOMPOSITION")
print("=" * 70)


subproblem1 = """
A farmer has 15 cows and buys 7 more.
How many cows does he have now?
Give only the numerical answer.
"""

subproblem2 = """
A farmer has 22 cows and sells half of them.
How many cows remain?
Give only the numerical answer.
"""


answer1 = ask_llm(subproblem1)
answer2 = ask_llm(subproblem2)

print("\nSubproblem 1:")
print(answer1)

print("\nSubproblem 2:")
print(answer2)


# ============================================================
# STEP 13: Final Analysis
# ============================================================

print("\n" + "=" * 70)
print("FINAL ANALYSIS")
print("=" * 70)

print("""
Direct Prompt:
The model receives only the problem and produces an answer.

Step-by-Step Prompt:
The model is explicitly asked to provide a brief calculation
summary before giving the final answer. This makes the arithmetic
easier to inspect.

Correct calculation:

15 + 7 = 22

22 / 2 = 11

Final answer = 11 cows.

Conclusion:
Both approaches can solve a simple arithmetic problem, but
explicitly requesting a concise calculation summary can make
the solution easier to verify.
""")

print("✅ Experiment 10 completed successfully.")
```
